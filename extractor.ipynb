```python
from pathlib import Path
import shutil
from docling.document_converter import DocumentConverter

#récupérer le chemin actuel
cwd = Path.cwd()
docs = cwd / "docs"

#Récupérer les sous répertoires de docs
for i, sub_dir in enumerate(docs.iterdir()):
    #Pour chaque sous répertoire, récupérer et convertir les fichiers en .md
    if sub_dir.is_dir():
        name = sub_dir.name
        print(f"## {i+1}, {name}\n")
        #Vérifier si les noms de répertoires contiennent des espaces et les remplacer par des underscores
        if " " in name:
            name = name.replace(" ", "_")
            sub_dir.rename(sub_dir.parent / name)

        
        if (sub_dir / "markdowns").exists():
            shutil.rmtree(sub_dir / "markdowns")
        #créer un répertoir markdown
        Path.mkdir(sub_dir / "markdowns")

        #récuperer les fichiers de chaque sous répertoire et les convertir en .md
        for j, file in enumerate(sub_dir.iterdir()):
            if file.name == ".DS_Store":
                continue
            #vérifier qu'on a bien affaire à un répertoire
            if file.is_file():
                print(f"** - {j+1} - {file.name}")
                #prendre le même nom que le fichier originel pour la version markdown
                markdown_name = file.name.replace(".", "_")
                markdown_name += ".md"
                
                #initialiser le convertisseur
                converter = DocumentConverter()
                result = converter.convert(file)

                #Créer le fichier markdown et éccrire la version transrcite à l'intérieur
                markdown = result.document.export_to_markdown()
                markdown_file = Path(sub_dir / "markdowns" / markdown_name)
                markdown_file.write_text(markdown, encoding="utf-8")

                print("\n\n")



```
